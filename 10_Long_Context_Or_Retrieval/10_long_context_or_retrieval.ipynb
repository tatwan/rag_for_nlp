**CPU | API key: `OPENAI_API_KEY` (chat and embedding calls) | Self-study**

## Goal

Decide between putting the whole corpus in the prompt and retrieving a few chunks, from answers, tokens, cost and
time measured on the same questions. The moment that teaches it: the whole corpus fits in one prompt, yet it answers
fewer questions right than retrieval, at many times the cost. The model answers the questions it missed when it gets
the right passage alone, but not when that passage sits inside the whole corpus.

Every lab so far has paid for retrieval: an index to build and keep current, a search before every answer, and in
Lab 08 a second index that added work for no gain. This corpus is small enough to fit in one prompt, so this lab
asks what retrieval buys here, in answers, tokens and dollars.

## What you will learn

- How to answer the same questions three ways (no evidence, three retrieved chunks, the whole corpus) and record the
  tokens and seconds of every answer.
- How to compare the three routes with one answer check, and read every answer it flags.
- How to price each route per 1,000 questions, and what a cached prefix changes (an OpenAI-only feature).
- How an answer changes when the passage that holds it gets more text around it, or moves.
- How to write a decision rule from your own table, and see where the corpus stops fitting in the prompt.

## Part A: Compare no evidence, retrieval and the whole corpus on two questions

Part A builds the three routes and runs them on two questions: one the model knows from training, and one it
cannot know.

| Route | What the model reads besides the question | What it needs first |
|---|---|---|
| No evidence | nothing: it answers from what it learned in training | nothing |
| Retrieval | the three chunks closest to the question, Lab 05's pipeline | an index of the 277 chunks |
| Whole corpus | every passage and policy, as `[ID] text` lines | nothing: no chunks, no index |

The whole corpus is the 120 SQuAD paragraphs and the five policies, each as one `[ID] text` evidence line, so the
model can cite it and Lab 02's quote check can read it. Count the tokens:

```python
import json
import tiktoken

DATA = ROOT / "data"
passages = json.loads((DATA / "squad-passages.json").read_text(encoding="utf-8"))
policies = json.loads((DATA / "support-case.json").read_text(encoding="utf-8"))
corpus_lines = [f"[{doc['source_id']}] {doc['text']}" for doc in passages + policies]

encoding = tiktoken.get_encoding(settings["TOKEN_ENCODING"])
corpus_tokens = len(encoding.encode("\n".join(corpus_lines)))
print(len(corpus_lines), "lines |", corpus_tokens, "tokens")
print(corpus_lines[0][:100], "...")
```

19,466 tokens: the paragraphs' text, plus an ID on each line and the policies. `gpt-4o-mini` reads up to
128,000 tokens in one request, so the whole corpus fits several times over. Part E comes back to that limit.

The retrieval route is Lab 05's pipeline from `shared/lc_pipeline.py`, which Lab 06 Part D prints: `build_store()` puts
the 277 chunks in an `InMemoryVectorStore`, `retrieve()` returns the top `k=3`, `evidence()` writes them as `[ID] text`
lines, and `answer()` asks the model for a cited answer.

The two routes read the same text, cut differently. The store holds 277 records: the 120 paragraphs cut into
272 smaller chunks (`S03-P00-C00` is chunk 0 of paragraph `S03-P00`), plus the five policies whole. So every chunk
retrieval can return is also inside the whole corpus's 125 lines. Build the store and retrieve for the Tesla
question:

```python
from shared import lc_pipeline

documents = lc_pipeline.to_documents(lc_pipeline.load_corpus())
store = lc_pipeline.build_store(documents, settings)
print(len(documents), "documents in the store")
question = "In what year was Nikola Tesla born?"
for line in lc_pipeline.evidence(lc_pipeline.retrieve(store, question)):
    print(len(encoding.encode(line)), "tokens |", line[:80], "...")
```

Three chunks, 275 tokens of evidence (73 + 102 + 100), where the whole-corpus route sends 19,466. Both
routes answer with Lab 05's `answer()`: the same rules and the same `CitedAnswer` schema, so only the evidence differs.
Read it:

```python
import inspect

print(inspect.getsource(lc_pipeline.answer))
```

`answer()` returns only the `CitedAnswer`. To price a route you also need the tokens billed. With
`include_raw=True`, `with_structured_output` returns a dictionary instead
([LangChain docs](https://docs.langchain.com/oss/python/langchain/models#structured-output)): `parsed`, the
`CitedAnswer`, and `raw`, the model's `AIMessage` with `usage_metadata`. `ask()` below is `answer()` with that one
change, and it keeps the evidence lines so the quotes can be checked later. A **callback**, an object LangChain calls
after every model call, adds up the tokens of every call in the notebook. It sits on `chat`, and `structured` is
built from `chat`, so it also counts every call `ask()` makes:

```python
from langchain_core.callbacks import UsageMetadataCallbackHandler
from shared.answer import RULES
from shared.settings import lc_chat_model

usage_total = UsageMetadataCallbackHandler()     # adds up the tokens of every call; Part E prints the total
chat = lc_chat_model("CHAT_MODEL", settings)
chat.callbacks = [usage_total]
method = "json_schema" if settings["JSON_MODE"] == "json_schema" else "json_mode"
structured = chat.with_structured_output(lc_pipeline.CitedAnswer, method=method, include_raw=True)

def ask(question, lines):
    """lc_pipeline.answer() with include_raw=True: the answer, its claims, the evidence and the tokens billed."""
    user = f"Question: {question}\nEvidence:\n" + "\n".join(lines)
    out = structured.invoke([("system", RULES), ("user", user)])
    return {"answer": out["parsed"].answer, "claims": [claim.model_dump() for claim in out["parsed"].claims],
            "lines": lines, "usage": out["raw"].usage_metadata}

record = ask(question, lc_pipeline.evidence(lc_pipeline.retrieve(store, question)))
print(record["answer"])
print(record["usage"])
```

The answer is the `CitedAnswer`'s text, and `usage` is what the API billed for this one call:

- `input_tokens`: 506, everything sent: the rules, the answer's schema, the question and the 275 tokens of
  evidence.
- `output_tokens`: 51, the JSON the model wrote: the answer and its claims.
- `input_token_details["cache_read"]`: input tokens read from OpenAI's prompt cache, 0 here. Part C comes back
  to it.

Now the three routes, as functions of the question. The no-evidence route gets a one-line instruction, since there is
no evidence to cite. `run()` times a route from the question to the answer, the retrieval route's search included.

```python
import time

def no_evidence(question):
    reply = chat.invoke([("system", "Answer the question in one sentence."), ("user", question)])
    return {"answer": reply.content, "claims": [], "lines": [], "usage": reply.usage_metadata}

def retrieval(question):
    return ask(question, lc_pipeline.evidence(lc_pipeline.retrieve(store, question)))

def whole_corpus(question):
    return ask(question, corpus_lines)

def run(route, question):
    start = time.perf_counter()
    record = route(question)
    record["seconds"] = time.perf_counter() - start
    return record

ROUTES = {"no evidence": no_evidence, "retrieval": retrieval, "whole corpus": whole_corpus}
```

Ask the Tesla question and one about Atlas, a product made up for these labs, so no model learned its policies in
training. Predict first: which route will get the Atlas question wrong, and how?

```python
for q in (question, "How long does an Atlas reset link remain valid?"):
    print(q)
    for name, route in ROUTES.items():
        record = run(route, q)
        print(f"  {name:13} {record['usage']['input_tokens']:6} tokens in | {record['seconds']:.2f} s | {record['answer']}")
```

All three routes give 1856 for Tesla, a famous fact the model learned in training. For Atlas, the no-evidence
route answers *An Atlas reset link typically remains valid for 24 hours.*: fluent and invented, since the policy says
30 minutes, and nothing in the reply says it is a guess. Retrieval and the whole corpus both say 30 minutes.

The input tokens show where each route's cost comes from:

- **No evidence, 26 or 28:** the one-line instruction and the question.
- **Retrieval, 506 or 296:** about 230 tokens of rules, schema and question, plus the three chunks it found:
  275 tokens for Tesla, and much less for Atlas, whose policies are one or two short sentences each. Retrieval's
  input depends on what it finds.
- **Whole corpus, about 19,700:** the same 230 tokens plus all 19,466 tokens of the corpus, whatever the question.

The seconds are single calls (1.17 to 2.10); Part C reads them over 34 questions.

## Part B: Compare the three routes on 34 questions and read every flagged answer

Two questions cannot show which kinds of question each route gets wrong. Part B runs the three routes on Lab 04's
questions: the 30 development questions with SQuAD's answer strings, and four about Atlas with the answers the
policies give.

```python
dev = json.loads((DATA / "development-questions.json").read_text(encoding="utf-8"))
questions = [{"question": q["question"], "expected": sorted({a["text"] for a in q["answers"]}),
              "support": q["known_support"]} for q in dev]
questions += [  # written for Lab 04: the answers the Atlas policies give
    {"question": "How long does an Atlas reset link remain valid?", "expected": ["30 minutes"], "support": ["H02"]},
    {"question": "Who can export all reports in Atlas?", "expected": ["workspace owners"], "support": ["H03"]},
    {"question": "Can temporary Atlas accounts reset by email?", "expected": ["support agent"], "support": ["H01"]},
    {"question": "Do I need the mobile app to reset my Atlas password?", "expected": ["no", "by email"], "support": ["H01"]},
]
print(len(questions), "questions")
```

Each question is a dictionary with three fields:

- `question`: the text sent to the model, and the only field it sees.
- `expected`: the accepted answers, which the screen looks for after the answer comes back. SQuAD gives several
  for some questions, so it is a list.
- `support`: the IDs of the sources that hold the answer (`S03-P00`, `H02`). Part D reads it.

One answer per question and route: 102 calls. `results` maps each route's name to a list of 34 records in the
order of `questions`, so `results["retrieval"][i]` answers `questions[i]`. Every cell below pairs them with
`zip`, which relies on that order.

```python
results = {name: [run(route, q["question"]) for q in questions] for name, route in ROUTES.items()}
print({name: len(records) for name, records in results.items()})
```

Check every answer with Lab 04's **screen**: it passes when one of the expected strings appears in the answer, as
whole words in lower case. `words()`, from `shared/answer.py`, lowers the case and keeps only the words, so
*30 Minutes.* becomes `30 minutes`. The spaces added around both texts make the match whole words: ` 30 minutes `
is found in ` links expire after 30 minutes `, but not in ` links expire after 130 minutes `. A screen flags
answers to read; it does not grade them.

A refusal is counted on its own. Lab 02's rules, used by retrieval and the whole corpus, end with: *If the evidence
does not answer the question, the answer is "Insufficient evidence." and the claims list is empty.*

```python
from shared.answer import quote_found, words

INSUFFICIENT = "Insufficient evidence."

def screen(answer_text, expected):
    answer = f" {words(answer_text)} "
    return any(f" {words(text)} " in answer for text in expected)

for name, records in results.items():
    passed = sum(screen(r["answer"], q["expected"]) for r, q in zip(records, questions))
    refused = sum(r["answer"].strip() == INSUFFICIENT for r in records)
    print(f"{name:13} passed {passed} of {len(questions)} | refused {refused}")
```

The no-evidence route refused nothing because its one-line instruction has no such rule, not because it knew the
answers.

A count says how many answers missed the expected words, not why: a flagged answer can be wrong, a refusal, or
right in other words. Print every answer the screen flags, route by route, and read them:

```python
for name, records in results.items():
    print(f"\n{name}")
    for r, q in zip(records, questions):
        if not screen(r["answer"], q["expected"]):
            print(f"  {q['question'][:60]:60} | expected {q['expected'][0][:25]!r} | {r['answer'][:90]}")
```

- The no-evidence route has 15 flagged: less famous facts missed from memory (*the Royal Łazienki Park*), and an
  invented policy for three of the four Atlas questions (*24 hours*, *administrators*, *Yes, ... by email*).
- Retrieval has the three answers Lab 04 traced: the complexity question names the field, the temporary-accounts
  question is refused, and the mobile-app question gets *Yes, password reset uses a mobile app.*, from H04, the policy of
  Boreal, a second made-up product.
- The whole corpus has 10: the same three (the temporary-accounts answer is now *Yes, ... can reset by email.*, the
  opposite of H01), five refusals of questions whose paragraph is in the prompt, the method question answered with
  the field's name, and *Norseman*, right in other words than the label's *Norseman, Viking* (Lab 04's incomplete
  label).

The retrieval route and the whole-corpus route read the same rules, and the whole corpus holds every chunk retrieval
found. **More evidence gave fewer answers.**

Last, the citations, checked with Lab 02's quote check (the no-evidence route has nothing to cite):

```python
for name, records in results.items():
    claims = [(r, claim) for r in records for claim in r["claims"]]
    found = sum(quote_found(claim, r["lines"]) for r, claim in claims)
    print(f"{name:13} claims {len(claims)} | quotes found {found}")
    for r, claim in claims:
        if not quote_found(claim, r["lines"]):
            print(f"  not found: [{claim['source_id']}] {claim['quote'][:90]}")
```

Retrieval made 35 claims, and all 35 quotes were found in the lines they cite; the whole corpus made 30, 29 found,
each citing a whole paragraph instead of a chunk. The one not found is printed: words from the Super Bowl paragraph
about Von Miller, cited to `S02-P02`, a paragraph about the Normans. The words are real, but not in the line the claim names,
and that is exactly what the check catches. Long context kept its citations checkable: written as `[ID] text` lines, the corpus can be cited and
checked like retrieved chunks. What it changed was the answers.

**Important:** a found quote says that the quoted words are in the line the claim cites. It does not say that
the line answers the question, or that the answer is right. That is why retrieval can have 35 quotes found and
31 answers passed: the two checks count different things.

## Part C: Price each route beside its answers, then cache the corpus

Part B said which route answers best. Part C puts the cost of each route beside its answers: tokens and dollars
per question, and seconds.

The prices come from `models.env`, in US dollars per million tokens. `per_question()` averages the input and output
tokens over the 34 records and prices one question:

$$
\text{dollars per question} = \frac{\text{input tokens} \times \text{input price} + \text{output tokens} \times \text{output price}}{1{,}000{,}000}
$$

The table shows it times 1,000. The screen count stays in the first column: a fast failure is always the cheapest
row.

```python
from shared.settings import chat_url

PRICE_IN, PRICE_OUT = float(settings["CHAT_PRICE_INPUT"]), float(settings["CHAT_PRICE_OUTPUT"])

def per_question(records):
    tokens_in = sum(r["usage"]["input_tokens"] for r in records) / len(records)
    tokens_out = sum(r["usage"]["output_tokens"] for r in records) / len(records)
    return tokens_in, tokens_out, (tokens_in * PRICE_IN + tokens_out * PRICE_OUT) / 1e6

print(f"{'route':13} {'passed':>6} {'in/question':>12} {'out/question':>13} {'$ per 1,000':>12}  seconds: median (range)")
for name, records in results.items():
    passed = sum(screen(r["answer"], q["expected"]) for r, q in zip(records, questions))
    tokens_in, tokens_out, dollars = per_question(records)
    seconds = sorted(r["seconds"] for r in records)
    print(f"{name:13} {passed:6} {tokens_in:12,.0f} {tokens_out:13,.0f} {1000 * dollars:12.2f}  "
          f"{seconds[len(seconds) // 2]:.2f} ({seconds[0]:.2f} to {seconds[-1]:.2f})")
print(f"\n{settings['CHAT_MODEL']} at {chat_url(settings)}; prices from models.env")
```

Retrieval passes 31 for 0.11 dollars per 1,000 questions; the whole corpus passes 24 for 2.99. With the prices in
`models.env` for `gpt-4o-mini` (0.15 dollars per million input tokens, 0.60 per million output tokens):

$$
\text{whole corpus: } \frac{19{,}700 \times 0.15 + 65 \times 0.60}{1{,}000{,}000} \times 1{,}000 = 2.99 \qquad \text{retrieval: } \frac{489 \times 0.15 + 69 \times 0.60}{1{,}000{,}000} \times 1{,}000 = 0.11
$$

The whole corpus's input is 40 times retrieval's (19,700 against 489 tokens a question), but its price only about 26
times: both routes write about the same answer, and output tokens cost four times as much as input tokens. The
output is about a third of retrieval's price and about 1 percent of the whole corpus's.

The seconds differ least: medians of 1.67 and 2.11 with overlapping ranges, from one run. The median is the
middle of the 34 sorted times, so one slow call does not move it. The no-evidence route is the cheapest and
fastest row (0.02 dollars, 0.96 seconds), and the one that invents policies.

Retrieval also pays for its index: the 277 chunks embedded once, well under a cent, and one short embedding per
question. The whole corpus needs no index, and pays for the corpus again on every question.

### A cached prefix (OpenAI only)

OpenAI caches the beginning of recent prompts. When a new request starts with the same tokens as a recent one, for
at least a minimum length (about 1,024 tokens; the guide gives it per model), those tokens are read from the cache
and billed at the cached price, `CHAT_PRICE_CACHED_INPUT`, half the input price for `gpt-4o-mini`
([prompt caching guide](https://developers.openai.com/api/docs/guides/prompt-caching)). LangChain reports
them as `cache_read` in `usage_metadata["input_token_details"]`; an endpoint that does not report them prints 0. How
much of Part B's whole-corpus input was cached?

```python
def cached(record):
    return record["usage"].get("input_token_details", {}).get("cache_read", 0)

whole = results["whole corpus"]
print(f"cached {sum(cached(r) for r in whole):,} of {sum(r['usage']['input_tokens'] for r in whole):,} input tokens")
```

19,584 of 669,799: one call's worth, the Atlas reset question, the one question Part A had already asked with
exactly the same prompt (Part B's Tesla question ends with a space, so its prompt differs). (Run the notebook
again within the hour and much more is cached: every prompt repeats.)

Why so little? The cache matches a prompt from its first token and stops at the first token that differs. With
the question first, two prompts start like this (`[rules]` is the system message):

- `[rules] Question: Who ruled the duchy of Normandy` `Evidence: [S00-P00] Super Bowl 50 was ...`
- `[rules] Question: In what year was Nikola Tesla born?` `Evidence: [S00-P00] Super Bowl 50 was ...`

They share only the rules and the schema, about 200 tokens, far below the 1,024 the cache needs. Put the corpus first, and
every question starts with the same 19,466 tokens; only the end differs:

- `[rules] Evidence: [S00-P00] Super Bowl 50 was ... [H05] ...` `Question: Who ruled the duchy of Normandy`
- `[rules] Evidence: [S00-P00] Super Bowl 50 was ... [H05] ...` `Question: In what year was Nikola Tesla born?`

`whole_corpus_first()` is `ask()` with that order:

```python
def whole_corpus_first(question):
    user = "Evidence:\n" + "\n".join(corpus_lines) + f"\nQuestion: {question}"
    out = structured.invoke([("system", RULES), ("user", user)])
    return {"answer": out["parsed"].answer, "claims": [claim.model_dump() for claim in out["parsed"].claims],
            "lines": corpus_lines, "usage": out["raw"].usage_metadata}

first = [run(whole_corpus_first, q["question"]) for q in questions]
print(f"cached {sum(cached(r) for r in first):,} of {sum(r['usage']['input_tokens'] for r in first):,} input tokens")
```

Price it at the cached price, and screen its answers: a change made for cost is still a change to the prompt.

`dollars()` prices each call's cached tokens at `CHAT_PRICE_CACHED_INPUT` and the rest at the full input price.
The table above priced every input token at the full price, so the question-first row below costs a little less than
the table's 2.99: its 19,584 cached tokens are priced at half.

```python
PRICE_CACHED = float(settings["CHAT_PRICE_CACHED_INPUT"])

def dollars(r):
    full = r["usage"]["input_tokens"] - cached(r)
    return (full * PRICE_IN + cached(r) * PRICE_CACHED + r["usage"]["output_tokens"] * PRICE_OUT) / 1e6

for name, records in [("question first", whole), ("corpus first", first)]:
    passed = sum(screen(r["answer"], q["expected"]) for r, q in zip(records, questions))
    refused = sum(r["answer"].strip() == INSUFFICIENT for r in records)
    print(f"{name:14} passed {passed} of 34 | refused {refused} | $ per 1,000 questions {1000 * sum(map(dollars, records)) / len(records):.2f}")
print("\nscreen result changed, with the corpus-first answer:")
for q, a, b in zip(questions, whole, first):
    before, after = screen(a["answer"], q["expected"]), screen(b["answer"], q["expected"])
    if before != after:
        print(f"  {'fail -> pass' if after else 'pass -> fail'} | {q['question'][:45]:45} | {b['answer'][:70]}")
```

With the corpus first, 626,688 of 669,798 input tokens were cached, more than nine tenths, and the cost per 1,000
questions fell from 2.95 to 1.61 dollars. The answers moved too: 23 passed instead of 24, 9 were refused instead of 5,
and 11 questions changed their screen result. **A total that moved by one hid eleven changes:**

- **fail -> pass, five:** the duchy, three complexity questions and the temporary accounts, which now gets the
  right *support agent*.
- **pass -> fail, six:** Curie, Pulaski, the Vistula's animals, Normandy's country, and Tesla's ethnicity and
  emigration, every one now *Insufficient evidence.* The new failures are all refusals of questions the corpus
  answers. On `gpt-4o-mini` the order of the prompt is part of the answer. Retrieval's prompts, about 500
tokens, are too short to cache and cheap without it.

## Part D: Test why the whole corpus loses answers its passage holds

Part D asks why the whole corpus lost answers that retrieval found. The passage that answers each of these
questions was in the whole-corpus prompt every time; what we can change is the text around it and where it sits.
First, the questions retrieval passed and the whole corpus failed, with the source that holds each answer:

```python
lost = [q for q, r, w in zip(questions, results["retrieval"], results["whole corpus"])
        if screen(r["answer"], q["expected"]) and not screen(w["answer"], q["expected"])]
for q in lost:
    print(q["support"], "|", q["question"])
```

Seven questions: four about the Normans (`S02-P01`, `S02-P02`), one about Tesla (`S03-P02`) and two about
complexity theory (`S04-P01`, `S04-P02`). One is not a real loss: the whole corpus answered *Norseman*, right in other
words. `support` is a list because a question can have more than one source; each of these has one, so the code
reads `[0]`.

A source ID names its place: `S02-P01` is article `S02`, paragraph `P01`. The three lines that start with `[S02` are
its article, which is how `contexts()` finds the second context below.

For each question, build six contexts around its supporting passage, the paragraph its label names. Each one
changes one thing from the one above it, so a change in the answers points to that thing:

| Context | What the model reads | What changed |
|---|---|---|
| passage alone | the supporting passage only, Lab 04's oracle context | the start: can the model answer from it at all? |
| its article | the three paragraphs of the same article | two related paragraphs added |
| whole corpus | all 125 lines, as in Part B | the other 122 lines added |
| moved first, to the middle, last | all 125 lines, with the passage taken out of its place and put first, 63rd or last | where the passage sits, and so which lines are next to it |

In the code, `line_of` maps each ID to its line: `line[1:line.index("]")]` is the text between `[` and `]`. `rest`
is the corpus without the passage, and `rest[:62] + [support] + rest[62:]` puts it back after the first 62 lines,
at line 63. The cell prints the six contexts for the first question:

```python
line_of = {line[1:line.index("]")]: line for line in corpus_lines}

def contexts(source_id):
    support = line_of[source_id]
    rest = [line for line in corpus_lines if line != support]
    return {"passage alone": [support],
            "its article": [line for line in corpus_lines if line.startswith(f"[{source_id[:3]}")],
            "whole corpus": corpus_lines,
            "moved first": [support] + rest,
            "moved to middle": rest[:62] + [support] + rest[62:],
            "moved last": rest + [support]}

for name, lines in contexts(lost[0]["support"][0]).items():
    print(f"{name:16} {len(lines):3} lines, passage at line {lines.index(line_of[lost[0]['support'][0]]) + 1}")
```

Every whole-corpus context holds 125 lines; only the passage's line number changes. In its own place, the first
question's passage is line 8, near the start.

Ask each question three times in each context (126 calls: 7 questions, 6 contexts, 3 runs; 84 of them with the
whole corpus). Predict first: which contexts will answer?

```python
from collections import Counter

answers = {}
for q in lost:
    for name, lines in contexts(q["support"][0]).items():
        answers[q["question"], name] = [ask(q["question"], lines)["answer"] for _ in range(3)]

for name in contexts(lost[0]["support"][0]):
    row = [sum(screen(a, q["expected"]) for a in answers[q["question"], name]) for q in lost]
    print(f"{name:16} passed {sum(row):2} of {3 * len(lost)} | per question {row}")
```

Each `per question` list follows the order of the seven questions above: in the *its article* row, the 0 is the
third question, the original meaning of *Norman*.

Read the answers, not only the counts. Each row below shows the most common of the three replies
(`Counter(replies).most_common(1)`). These replies often repeat the question first and put the answer last, so a
reply longer than 85 characters is shown by its end: `...` and its last 82 characters. Read the count beside it:
in a 2-of-3 row the reply shown may be one that failed:

```python
for q in lost:
    print(f"\n{q['question']}  (expected {q['expected'][0]!r})")
    for name in contexts(q["support"][0]):
        replies = answers[q["question"], name]
        passed = sum(screen(a, q["expected"]) for a in replies)
        reply = Counter(replies).most_common(1)[0][0]
        print(f"  {name:16} passed {passed} of 3 | {reply if len(reply) <= 85 else '...' + reply[-82:]}")
```

Alone, the passage answers in 21 of 21 runs; inside its article in 18 of 21 (the misses are *Norseman or Northman*,
the label again); inside the whole corpus in 0 of 21, mostly *Insufficient evidence.* Only the text around it changed.

Moving the passage changed the answers again: first in the prompt 21 of 21, last 15 of 21, in the middle 9 of 21.
Their own places, near the start (line 8 for the first question), gave 0, so neither "near the start is safe" nor "the
middle is lost" holds. Moved to the middle or to the end, both complexity questions fail. Moved last, the method
question names the wrong field (*... is computational complexity theory.*); the other three of those rows end by
repeating the question, so their answer is in the part not shown.

**Important:** moving a passage changes two things at once: its position, and its neighbours. In its own place it
sits beside the other paragraphs of its article; moved, it does not. These runs cannot say which of the two
mattered.

Research has names for what Part D tests. **Lost in the middle** (Liu et al., 2023): models answered best when the
relevant passage sat at the start or the end of a long prompt, and worst in the middle. **Context rot** (Chroma,
2025): answers get worse as the input grows, and related text that does not answer makes it worse. Our run shows
some of both and neither as a rule: the middle (9 of 21) did worse than first (21) or last (15), yet the passage's
own place near the start gave 0. Whether an answer survives a long prompt depends on where its passage sits and
what surrounds it, and only a run on your questions, on your model, shows it. Retrieval sidesteps the question: its
prompt is three chunks, the closest one first, so there is little to bury the passage in.

## Part E: Find where the whole corpus stops fitting, and write the decision rule

Part E turns the measurements into a rule. The whole-corpus route's tokens grow with the corpus; retrieval sends
three chunks at any size. From Part B's averages, the cost as the corpus grows. The table multiplies the corpus by
1, 5, 10 and 100, keeps the rest of the prompt (`overhead`, about 234 tokens: the rules, the question and the
schema) and the output as they were, and prices every input token at the full price; `/ 1e3` turns a price per
million tokens into a price per 1,000 questions:

```python
ret_in, ret_out, _ = per_question(results["retrieval"])
whole_in, whole_out, _ = per_question(results["whole corpus"])
overhead = whole_in - corpus_tokens            # the rules, the question and the schema

print(f"{'corpus size':>12} {'whole corpus: in/question':>26} {'$ per 1,000':>12} | retrieval: $ per 1,000")
for times in (1, 5, 10, 100):
    tokens = times * corpus_tokens + overhead
    print(f"{times:>11}x {tokens:26,.0f} {(tokens * PRICE_IN + whole_out * PRICE_OUT) / 1e3:12.2f} |"
          f" {(ret_in * PRICE_IN + ret_out * PRICE_OUT) / 1e3:.2f}")
```

Retrieval's column does not move: three chunks cost 0.11 dollars per 1,000 questions at any corpus size. The whole
corpus's price grows with the corpus: 14.67 dollars at five times this corpus, 292.06 at a hundred times.

The table prices tokens only. It does not say how well either route answers at that size: Part D found that more
text around a passage can lose its answer, and a larger corpus also gives retrieval more near misses to choose
from. That takes a new run, not this table.

The model has a limit on what one request can hold, its **context window**. Stand in for a corpus seven times larger
by repeating the lines seven times, and ask the Tesla question. `ContextOverflowError` is LangChain's error for a
prompt longer than the window, whatever the provider; the cell prints it instead of stopping:

```python
from langchain_core.exceptions import ContextOverflowError

bigger = corpus_lines * 7
print(len(encoding.encode("\n".join(bigger))), "tokens")
try:
    print(ask(question, bigger)["answer"])
except ContextOverflowError as error:
    print(type(error).__name__, "|", error)
```

136,262 tokens by the local count, and the API refused the request: *This model's maximum context length is
128000 tokens.* The API counted 136,492: the same lines plus the rules, the question and the schema (the error
names 107 tokens of schema). Five copies of this corpus still fit (97,564 tokens a question in the table above); seven do not. The limit sits at about six and a
half copies: (128,000 − 234) / 19,466 ≈ 6.6. Long
before that, the price grows: 14.67 dollars per 1,000 questions at five times this corpus, against 0.11. A larger
window moves the limit; it does not lower the cost.

Put the measurements together. The rule below is written from this run on `gpt-4o-mini`; write yours from your own
table, on your model:

| If | Choose | What this run measured |
|---|---|---|
| The answer is private or recent, or must name its source | not the no-evidence route | 19 of 34 passed; three of the four Atlas answers invented; nothing to cite |
| The corpus fits in the window, questions are few, and you have measured the answers on your model | the whole corpus can be an option; put it before the question so it is cached | 24 of 34 at 2.99 dollars per 1,000 (1.61 cached, 23 passing); lost questions 0 of 21 runs |
| The corpus is large or growing, questions are many, or some documents must be left out for some users | retrieval | 31 of 34 at 0.11 dollars per 1,000, three chunks a question at any corpus size |

A window size decides neither of the two things that matter most: how well your model answers from a long prompt,
and whether some documents must be left out, which takes a search with a filter (Lab 05, Part G).
